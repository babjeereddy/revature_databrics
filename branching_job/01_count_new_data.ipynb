```python
from pyspark.sql import functions as F

dbutils.widgets.text("process_date", "")  # yyyy-MM-dd, blank means today
p = dbutils.widgets.get("process_date").strip()

if p == "":
    process_date_expr = F.current_date()
else:
    process_date_expr = F.to_date(F.lit(p))

df = spark.table("project.bronze.orders").where(F.col("ingest_date") == process_date_expr)
new_count = df.count()

dbutils.jobs.taskValues.set(key="new_count", value=new_count)

```
