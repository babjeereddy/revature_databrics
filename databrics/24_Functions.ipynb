```sql
CREATE OR REPLACE FUNCTION get_customers(p_city STRING)
RETURNS TABLE (customer_id STRING, customer_name STRING , city STRING)
RETURN
  SELECT customer_id, customer_name, city
  FROM project.bronze.customers
  WHERE city = p_city;
  
```

```sql
update customers set city='Chennai';
```

```sql
SELECT c.customer_id, c.customer_name FROM get_customers('Chennai') AS c;
```

```sql
CREATE or REPLACE FUNCTION mask_pan(pan STRING)
RETURNS STRING
LANGUAGE PYTHON
AS $$
return pan[:2] + "XXXX" + pan[-2:]
$$;
```
