```sql
%python
from pyspark.sql.functions import to_date
merge_data = [
    (1, "Ravi", 51000.0, "IT", "2024-01-01"),
    (7, "Divya", 48000.0, "HR", "2024-02-15")
]

df_merge = spark.createDataFrame(merge_data, ["id", "name", "salary", "dept", "doj"])
df_merge = df_merge.withColumn("doj", to_date("doj"))
df_merge.createOrReplaceTempView("staging_emp")
```

```sql
select * from staging_emp
```

```sql
MERGE INTO project.silver.emp_delta AS target
USING staging_emp AS source
ON target.id = source.id
WHEN MATCHED THEN
  UPDATE SET
    target.name   = source.name,
    target.salary = source.salary,
    target.dept   = source.dept
WHEN NOT MATCHED THEN
  INSERT (id, name, salary, dept, doj)
  VALUES (source.id, source.name, source.salary, source.dept, source.doj);

```
