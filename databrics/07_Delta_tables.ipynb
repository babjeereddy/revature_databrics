```sql

CREATE OR REPLACE TEMP VIEW emp_updates AS
SELECT * FROM VALUES
  (1, 'Ravi',  'IT',  55000),  
  (3, 'John',  'FIN', 48000),  
  (4, 'Meera', 'IT',  62000)   
AS t(id, name, dept, salary);
```

```sql
select * from emp_updates;
```

```sql
MERGE INTO project.silver.emp_delta AS target
USING emp_updates AS source
ON target.id = source.id

WHEN MATCHED THEN
  UPDATE SET
    target.name   = source.name,
    target.dept   = source.dept,
    target.salary = source.salary

WHEN NOT MATCHED THEN
  INSERT (id, name, dept, salary)
  VALUES (source.id, source.name, source.dept, source.salary);
```

```sql
INSERT INTO project.silver.emp_delta (id, name, dept, salary, status) VALUES
  (1, 'Ravi',  'IT',  50000, 'ACTIVE'),
  (2, 'Priya', 'HR',  60000, 'ACTIVE'),
  (3, 'John',  'FIN', 45000, 'ACTIVE');
```

```sql
select * from project.silver.emp_delta;
```

```sql
CREATE OR REPLACE TEMP VIEW emp_changes AS
SELECT * FROM VALUES
  (1, 'Ravi',  'IT',  55000, 'ACTIVE'),    -- existing → UPDATE
  (2, 'Priya', 'HR',  60000, 'INACTIVE'),  -- existing → DELETE
  (4, 'Meera', 'IT',  62000, 'ACTIVE')     -- new      → INSERT
AS t(id, name, dept, salary, status);
```

```sql
MERGE INTO project.silver.emp_delta       AS target
USING emp_changes          AS source
ON target.id = source.id


WHEN MATCHED AND source.status = 'INACTIVE' THEN
  DELETE

WHEN MATCHED AND source.status = 'ACTIVE' THEN
  UPDATE SET
    target.name   = source.name,
    target.dept   = source.dept,
    target.salary = source.salary,
    target.status = source.status

-- If not matched (new id) → INSERT
WHEN NOT MATCHED THEN
  INSERT (id, name, dept, salary, status)
  VALUES (source.id, source.name, source.dept, source.salary, source.status);
```

```sql
CREATE TABLE project.silver.dim_customer (
  cust_sk        BIGINT GENERATED ALWAYS AS IDENTITY,  
  cust_id        INT,       
  cust_name      STRING,
  city           STRING,
  phone          STRING,
  effective_from DATE,
  effective_to   DATE,
  is_current     BOOLEAN
)
USING DELTA;
```

```sql
desc project.silver.dim_customer;
```

```sql
CREATE OR REPLACE TEMP VIEW src_customer AS
SELECT * FROM VALUES
  (101, 'Ravi',   'Chennai', '9000000011'),
  (102, 'Priya',  'Bangalore', '9000000022'),
  (103, 'John',   'Mumbai',  '9000000033')
AS t(cust_id, cust_name, city, phone);
```

```sql
select * from src_customer;
```

```sql
INSERT INTO project.silver.dim_customer (cust_id, cust_name, city, phone, effective_from, effective_to, is_current)
SELECT
  cust_id,
  cust_name,
  city,
  phone,
  current_date()      AS effective_from,
  DATE '9999-12-31'   AS effective_to,
  true                AS is_current
FROM src_customer;

```

```sql
select * from project.silver.dim_customer;
```

```sql
CREATE OR REPLACE TEMP VIEW src_customer AS
SELECT * FROM VALUES
  (101, 'Ravi',  'Hyderabad', '9000000011'),  
  (102, 'Priya', 'Bangalore', '9000000022'),  
  (104, 'Meera', 'Chennai',   '9000000044')   
AS t(cust_id, cust_name, city, phone);
```

```sql
MERGE INTO project.silver.dim_customer AS tgt
USING src_customer  AS src
ON tgt.cust_id = src.cust_id
AND tgt.is_current = true      

WHEN MATCHED AND 
  (
    tgt.cust_name <> src.cust_name OR
    tgt.city      <> src.city      OR
    tgt.phone     <> src.phone
  )
THEN UPDATE SET
  tgt.effective_to = current_date(),   -- expire old row
  tgt.is_current   = false

WHEN NOT MATCHED THEN
  INSERT (
    cust_id, cust_name, city, phone,
    effective_from,      effective_to,     is_current
  )
  VALUES (
    src.cust_id, src.cust_name, src.city, src.phone,
    current_date(),              DATE '9999-12-31', true
  );

```
